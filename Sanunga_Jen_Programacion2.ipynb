1.- Diseño de la clase

Cree una clase llamada AnalizadorDatos con los siguientes elementos:
• Atributo: datos (una lista de números).

• Métodos:

obtener_maximo() → devuelve el valor máximo.

obtener_minimo() → devuelve el valor mínimo.

calcular_media() → devuelve el promedio de los datos.

mostrar_resumen() → imprime todos los valores anteriores.

```python
class AnalizadorDatos:
    def __init__(self, datos):
        self.datos = datos

    def obtener_maximo(self):
        return max(self.datos)

    def obtener_minimo(self):
        return min(self.datos)

    def calcular_media(self):
        return sum(self.datos) / len(self.datos)

    def mostrar_resumen(self):
        print(f"Máximo: {self.obtener_maximo()}")
        print(f"Mínimo: {self.obtener_minimo()}")
        print(f"Media: {self.calcular_media()}")
```

3.-Extensión: lectura dinámica

a) Añada a la clase un método agregar_dato(nuevo_valor) que permita añadir un nuevo número a la lista.

b) Permita que el usuario ingrese varios valores por teclado hasta
escribir “fin”.

c) Luego, muestre nuevamente el resumen con los nuevos datos
agregados.

```python
#a) Añada a la clase un método agregar_dato(nuevo_valor)
class AnalizadorDatos:
    def __init__(self, datos):
        self.datos = datos

    def obtener_maximo(self):
        return max(self.datos)

    def obtener_minimo(self):
        return min(self.datos)

    def calcular_media(self):
        return sum(self.datos) / len(self.datos)
        #añadir un nuevo dato
    def agregar_dato(self, nuevo_valor):
        self.datos.append(nuevo_valor)

    def mostrar_resumen(self):
        print(f"Maximo: {self.obtener_maximo()}")
        print(f"Minimo: {self.obtener_minimo()}")
        print(f"Media: {self.calcular_media()}")
        print(f"Rango: {self.obtener_maximo() - self.obtener_minimo()}")
```

```python
#b) Permita que el usuario ingrese varios valores por teclado hasta escribir "fin"

class AnalizadorDatos:
    def __init__(self, datos):
        self.datos = datos

    def obtener_maximo(self):
        return max(self.datos)

    def obtener_minimo(self):
        return min(self.datos)

    def calcular_media(self):
        return sum(self.datos) / len(self.datos)
        #añadir un nuevo dato
    def agregar_dato(self, nuevo_valor):
        self.datos.append(nuevo_valor)


    def mostrar_resumen(self):
        print(f"Maximo: {self.obtener_maximo()}")
        print(f"Minimo: {self.obtener_minimo()}")
        print(f"Media: {self.calcular_media()}")
        print(f"Rango: {self.obtener_maximo() - self.obtener_minimo()}")

analizador = AnalizadorDatos([])

while True:
    entrada = input("Ingrese un número (o 'fin' para terminar): ")
    if entrada.lower() == "fin":
        break
    else:
        analizador.agregar_dato(int(entrada))
```

```python
analizador.mostrar_resumen()
```

```python
#c) Luego, muestre nuevamente el resumen con los nuevos datos agregados.

print("Cantidad de datos" , len(analizador.datos))
print("Datos", analizador.datos)
print ("Resumen acutalizado")
analizador.mostrar_resumen()
```

4. Reflexión final
• ¿Cómo ayuda la orientación a objetos a organizar este tipo de
programas?

La orientación a objetos ayuda a organizar los programas porque permite dividir el código en partes más pequeñas llamadas clases y objetos. Cada clase tiene una función específica, lo que hace que el programa sea más ordenado, fácil de entender y de mantener.

• ¿Qué ventajas tiene encapsular las operaciones estadísticas dentro
de una clase?

Encapsular las operaciones estadísticas dentro de una clase tiene varias ventajas:

Organización: Todas las funciones relacionadas (media, moda, varianza, etc.) están en un solo lugar.

Reutilización: La clase se puede usar en otros programas sin necesidad de volver a escribir el código.

Seguridad: Se protege la información interna, evitando modificaciones incorrectas.

Facilidad de mantenimiento: Si hay errores o mejoras, se hacen en un solo sitio.

Claridad: El código se vuelve más limpio y fácil de entender.
