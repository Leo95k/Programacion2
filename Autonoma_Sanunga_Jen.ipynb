a) Cree una clase base ModeloIA con los métodos:

• entrenar(datos)

• predecir(entrada)

• evaluar(resultados_reales, resultados_predichos)

```python
class ModeloIA():
    def __init__(self , nombre): 
        self.nombre= nombre
        
    def entrenar(self, datos):
        print(f"Entrenar los datos ") 
        self.nombre = datos

    def predecir(self , entrada):
        self.nombre= entrada 
        print(f"Generando la prediccion en la entrada  de los datos")

    def evaluar(self, resultados):
        self.nombre = resultados
        print (f"Evaluar los resultados de {resultados}") 
```

1.2.3 Métodos Mágicos y Sobrecarga de Operadores

a) Implemente en la clase base ModeloIA los métodos especiales:

• __str__: devuelve una descripción legible del modelo.

• __add__: combina dos modelos y devuelve un nuevo modelo 
“fusionado”.

• __sub__: representa la diferencia conceptual entre modelos (por 
ejemplo, restar precisión simulada).

```python
class ModeloIA():
    def __init__(self , nombre , presicion):   
        self.nombre= nombre
        self.presicion= presicion 
        
    def __str__(self):
        return f"ModeloIA : {self.nombre} | Presicion actual: {self.presicion}"
    
    def __add__(self , otro ):
        nuevo_nombre= f"{self.nombre} + {otro.nombre}"
        nueva_exactitud= (self.presicion + otro.presicion) / 2
        print(f"Fusionando {self.nombre} con {otro.nombre}")
        return ModeloIA(nuevo_nombre , nueva_exactitud)
    
    def __sub__(self, otro):
        print(f" Calculando la diferencia entre {self.nombre} y {otro.nombre}")
        diferencia = abs (self.presicion - otro.presicion)
        return diferencia
    
    def predecir(self , entrada):
        return f"Predeccion de {self.nombre} con datos {entrada}" 
```

b) Cree ejemplos donde se sumen y resten modelos e imprima sus

```python
modelo_a= ModeloIA( "x" , 0.9)
modelo_b= ModeloIA("y" , 0.7)
```

```python
print(modelo_a)
print(modelo_b) 
```

```python
modelo_fusionado = modelo_a + modelo_b 
print(f"El resultado de la suma de los modelos es {modelo_fusionado}")
```

1.2.4 Patrón Singleton y Gestión de Sesiones de Entrenamiento

a) Cree una clase GestorSesion que controle las sesiones de 
entrenamiento usando el patrón Singleton.

• Use el método __new__ para garantizar una única instancia.

• Agregue métodos iniciar_sesion() y cerrar_sesion() que impriman 
mensajes simulando conexión/desconexión a un servidor de IA.

```python
class Gestor_sesion():
    _instancia= None

    def __new__(cls):
        if cls._instancia is None : 
            print(f"Creando una nueva instancia para gestionar la sesion")
            cls._instancia= super().__new__(cls) 

        return cls._instancia
            
    def iniciar_sesion(self):
        print(f"Servidor de IA : iniciando correctamente")
    
    def cerrar_sesion(self):
        print(f"Servidor de IA: cerrando sesion ")
```

```python
servi_1= Gestor_sesion()
servi_2= Gestor_sesion()
```

```python
#a
servi_1.iniciar_sesion() 
```

b) Verifique que, al crear varias instancias, todas sean el mismo objeto 
(id() igual).

```python
print(f" Id del servidor 1 : {id(servi_1)}")
print(f" Id del servidor 2 : {id(servi_2)}")

if id(servi_1) == id(servi_2):
    print(f"¡Correcto! ambas variables son los mismos objetos")
    
servi_2.cerrar_sesion()
```
